```python
import os
import pandas as pd

base_dir = "/content/humaid/events_set1"

train_data = []
dev_data = []
test_data = []

for root, dirs, files in os.walk(base_dir):
    for file in files:

        if not file.endswith(".tsv"):
            continue

        file_path = os.path.join(root, file)

        df_temp = pd.read_csv(file_path, sep="\t")

        if file.endswith("_train.tsv"):
            train_data.append(df_temp)

        elif file.endswith("_dev.tsv"):
            dev_data.append(df_temp)

        elif file.endswith("_test.tsv"):
            test_data.append(df_temp)

# Combine datasets
train_df = pd.concat(train_data, ignore_index=True)
dev_df = pd.concat(dev_data, ignore_index=True)
test_df = pd.concat(test_data, ignore_index=True)

print("Dataset combination completed!")
print("=" * 60)

print("Train shape:", train_df.shape)
print("Dev shape:", dev_df.shape)
print("Test shape:", test_df.shape)

print("\nTotal samples:", len(train_df) + len(dev_df) + len(test_df))
```

```python
print("Train:")
display(train_df.head())

print("\nDev:")
display(dev_df.head())

print("\nTest:")
display(test_df.head())
```

```python
from sklearn.preprocessing import LabelEncoder

# Create label encoder
label_encoder = LabelEncoder()

# Fit ONLY on training labels
label_encoder.fit(train_df["class_label"])

# Convert labels to numbers
train_df["label"] = label_encoder.transform(train_df["class_label"])
dev_df["label"] = label_encoder.transform(dev_df["class_label"])
test_df["label"] = label_encoder.transform(test_df["class_label"])

# Display mapping
print("Label Mapping:")
print("=" * 50)

for i, label in enumerate(label_encoder.classes_):
    print(f"{i}: {label}")
```

```python
import numpy as np
from sklearn.metrics import accuracy_score, precision_recall_fscore_support

def compute_metrics(eval_pred):
    predictions, labels = eval_pred

    predictions = np.argmax(predictions, axis=1)

    accuracy = accuracy_score(labels, predictions)

    precision, recall, f1, _ = precision_recall_fscore_support(
        labels,
        predictions,
        average="weighted",
        zero_division=0
    )

    return {
        "accuracy": accuracy,
        "precision": precision,
        "recall": recall,
        "f1": f1
    }

print("Metrics function created!")
```

```python
from transformers import TrainingArguments

training_args = TrainingArguments(
    output_dir="/content/humaid_distilbert",

    # Training
    num_train_epochs=3,
    per_device_train_batch_size=16,
    per_device_eval_batch_size=32,

    # Learning
    learning_rate=2e-5,
    weight_decay=0.01,

    # Evaluation
    eval_strategy="epoch",
    save_strategy="epoch",

    # Best model
    load_best_model_at_end=True,
    metric_for_best_model="f1",
    greater_is_better=True,

    # Logging
    logging_steps=100,

    # Performance
    fp16=True,

    # Avoid unnecessary external reporting
    report_to="none"
)

print("Training arguments configured!")
```

```python
from transformers import Trainer

trainer = Trainer(
    model=model,
    args=training_args,

    train_dataset=tokenized_dataset["train"],
    eval_dataset=tokenized_dataset["validation"],

    processing_class=tokenizer,
    data_collator=data_collator,

    compute_metrics=compute_metrics
)

print("Trainer created successfully!")
```

```python
import torch

def predict_humaid(text):

    inputs = tokenizer(
        text,
        return_tensors="pt",
        truncation=True,
        max_length=128
    )

    inputs = {
        key: value.to(model.device)
        for key, value in inputs.items()
    }

    model.eval()

    with torch.no_grad():
        outputs = model(**inputs)

    probabilities = torch.softmax(outputs.logits, dim=-1)

    predicted_id = torch.argmax(probabilities, dim=-1).item()

    confidence = probabilities[0][predicted_id].item()

    predicted_class = label_encoder.classes_[predicted_id]

    return predicted_class, confidence
```

```python
def predict_top3(text):

    inputs = tokenizer(
        text,
        return_tensors="pt",
        truncation=True,
        max_length=128
    )

    inputs = {
        key: value.to(model.device)
        for key, value in inputs.items()
    }

    model.eval()

    with torch.no_grad():
        outputs = model(**inputs)

    probabilities = torch.softmax(outputs.logits, dim=-1)[0]

    top_values, top_indices = torch.topk(probabilities, 3)

    print("Tweet:")
    print(text)
    print("\nTop 3 predictions:")

    for probability, index in zip(top_values, top_indices):
        label = label_encoder.classes_[index.item()]
        print(f"{label}: {probability.item():.2%}")
```

```python
from transformers import TrainingArguments

training_args = TrainingArguments(
    output_dir="/content/humaid_distilbert",

    num_train_epochs=3,

    per_device_train_batch_size=16,
    per_device_eval_batch_size=32,

    learning_rate=2e-5,
    weight_decay=0.01,

    eval_strategy="epoch",
    save_strategy="epoch",

    load_best_model_at_end=True,
    metric_for_best_model="f1",
    greater_is_better=True,

    logging_steps=100,

    fp16=True,

    report_to="none"
)

print("Training configuration ready!")
```

```python
import numpy as np
from sklearn.metrics import accuracy_score
from sklearn.metrics import precision_recall_fscore_support

def compute_metrics(eval_pred):

    predictions, labels = eval_pred

    predictions = np.argmax(predictions, axis=1)

    accuracy = accuracy_score(labels, predictions)

    precision, recall, f1, _ = precision_recall_fscore_support(
        labels,
        predictions,
        average="weighted",
        zero_division=0
    )

    return {
        "accuracy": accuracy,
        "precision": precision,
        "recall": recall,
        "f1": f1
    }

print("Metrics ready!")
```

```python
from transformers import Trainer
from transformers import DataCollatorWithPadding

data_collator = DataCollatorWithPadding(
    tokenizer=tokenizer
)

trainer = Trainer(
    model=model,
    args=training_args,

    train_dataset=tokenized_dataset["train"],
    eval_dataset=tokenized_dataset["validation"],

    processing_class=tokenizer,

    data_collator=data_collator,

    compute_metrics=compute_metrics
)

print("Trainer ready!")
```

```python
print("Starting HumAID training...")

train_result = trainer.train()

print("Training completed!")
```

```python
import torch

def predict_humaid(text):

    inputs = tokenizer(
        text,
        return_tensors="pt",
        truncation=True,
        max_length=128
    )

    inputs = {
        key: value.to(model.device)
        for key, value in inputs.items()
    }

    model.eval()

    with torch.no_grad():
        outputs = model(**inputs)

    probabilities = torch.softmax(
        outputs.logits,
        dim=-1
    )

    predicted_id = torch.argmax(
        probabilities,
        dim=-1
    ).item()

    confidence = probabilities[0][predicted_id].item()

    predicted_class = label_encoder.classes_[predicted_id]

    return predicted_class, confidence
```

```python
# ============================================
# STEP 6B: TRAINING CONFIGURATION
# ============================================

import numpy as np
from sklearn.metrics import accuracy_score, precision_recall_fscore_support
from transformers import TrainingArguments, Trainer

def compute_metrics(eval_pred):
    logits, labels = eval_pred

    predictions = np.argmax(logits, axis=-1)

    precision, recall, f1, _ = precision_recall_fscore_support(
        labels,
        predictions,
        average="weighted",
        zero_division=0
    )

    accuracy = accuracy_score(labels, predictions)

    return {
        "accuracy": accuracy,
        "precision": precision,
        "recall": recall,
        "f1": f1
    }

training_args = TrainingArguments(
    output_dir="/content/crisismmd_distilbert",

    # Evaluation and saving
    eval_strategy="epoch",
    save_strategy="epoch",

    # Training
    learning_rate=2e-5,
    per_device_train_batch_size=16,
    per_device_eval_batch_size=32,
    num_train_epochs=3,

    # Regularization
    weight_decay=0.01,

    # Logging
    logging_steps=100,
    report_to="none",

    # Best model
    load_best_model_at_end=True,
    metric_for_best_model="f1",
    greater_is_better=True,

    # GPU
    fp16=torch.cuda.is_available()
)

print("Training configuration created!")
```

```python
# ============================================================
# STEP 4: ENCODE CRISISMMd LABELS
# ============================================================

from sklearn.preprocessing import LabelEncoder

label_encoder = LabelEncoder()

crisis_model_df["label"] = label_encoder.fit_transform(
    crisis_model_df["class_label"]
)

print("========================================")
print("CRISISMMd LABEL MAPPING")
print("========================================")

for i, label_name in enumerate(label_encoder.classes_):
    print(f"{i}: {label_name}")

print("\nNumber of classes:", len(label_encoder.classes_))

print("\nEncoded label distribution:")
print(crisis_model_df["label"].value_counts().sort_index())
```

```python
# ============================================================
# STEP 5: TRAIN / VALIDATION / TEST SPLIT
# ============================================================

from sklearn.model_selection import train_test_split

# First split: 70% train, 30% temporary
train_crisis, temp_crisis = train_test_split(
    crisis_model_df,
    test_size=0.30,
    random_state=42,
    stratify=crisis_model_df["label"]
)

# Second split: temporary into validation and test
dev_crisis, test_crisis = train_test_split(
    temp_crisis,
    test_size=0.50,
    random_state=42,
    stratify=temp_crisis["label"]
)

print("========================================")
print("CRISISMMd SPLIT COMPLETED")
print("========================================")

print("Train:", train_crisis.shape)
print("Validation:", dev_crisis.shape)
print("Test:", test_crisis.shape)

print("\nTotal:",
      len(train_crisis) +
      len(dev_crisis) +
      len(test_crisis))

print("\nTRAIN labels:")
print(train_crisis["class_label"].value_counts())

print("\nVALIDATION labels:")
print(dev_crisis["class_label"].value_counts())

print("\nTEST labels:")
print(test_crisis["class_label"].value_counts())
```

```python
# ============================================================
# STEP 6: TOKENIZE CRISISMMd DATA
# ============================================================

from datasets import Dataset
from transformers import AutoTokenizer

print("Loading DistilBERT tokenizer...")

MODEL_NAME = "distilbert-base-uncased"

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

print("Tokenizer loaded successfully!")

# Keep only the columns needed for Hugging Face
train_text = train_crisis[["tweet_text", "label"]].copy()
val_text = dev_crisis[["tweet_text", "label"]].copy()
test_text = test_crisis[["tweet_text", "label"]].copy()

# Convert pandas -> Hugging Face Dataset
train_dataset = Dataset.from_pandas(
    train_text,
    preserve_index=False
)

val_dataset = Dataset.from_pandas(
    val_text,
    preserve_index=False
)

test_dataset = Dataset.from_pandas(
    test_text,
    preserve_index=False
)

# Tokenization function
def tokenize_function(examples):
    return tokenizer(
        examples["tweet_text"],
        truncation=True,
        padding="max_length",
        max_length=128
    )

print("\nTokenizing training data...")
train_tokenized = train_dataset.map(
    tokenize_function,
    batched=True,
    remove_columns=["tweet_text"]
)

print("Tokenizing validation data...")
val_tokenized = val_dataset.map(
    tokenize_function,
    batched=True,
    remove_columns=["tweet_text"]
)

print("Tokenizing test data...")
test_tokenized = test_dataset.map(
    tokenize_function,
    batched=True,
    remove_columns=["tweet_text"]
)

# Rename label -> labels for Transformers
train_tokenized = train_tokenized.rename_column("label", "labels")
val_tokenized = val_tokenized.rename_column("label", "labels")
test_tokenized = test_tokenized.rename_column("label", "labels")

# Set PyTorch format
train_tokenized.set_format("torch")
val_tokenized.set_format("torch")
test_tokenized.set_format("torch")

print("\n========================================")
print("TOKENIZATION COMPLETED")
print("========================================")

print("Train:", train_tokenized)
print("Validation:", val_tokenized)
print("Test:", test_tokenized)
```

```python
# ============================================================
# STEP 7: TRAIN CRISISMMD DISTILBERT
# ============================================================

import numpy as np
import torch

from transformers import TrainingArguments, Trainer
from sklearn.metrics import (
    accuracy_score,
    precision_recall_fscore_support
)

print("Preparing CrisisMMD training...")


# ============================================================
# 1. USE YOUR EXISTING TOKENIZED DATA
# ============================================================

train_data = train_dataset
val_data = val_dataset
test_data = test_dataset

print("Train:", len(train_data))
print("Validation:", len(val_data))
print("Test:", len(test_data))


# ============================================================
# 2. REMOVE tweet_text
# ============================================================

if "tweet_text" in train_data.column_names:
    train_data = train_data.remove_columns(["tweet_text"])

if "tweet_text" in val_data.column_names:
    val_data = val_data.remove_columns(["tweet_text"])

if "tweet_text" in test_data.column_names:
    test_data = test_data.remove_columns(["tweet_text"])


print("\nTraining columns:")
print(train_data.column_names)


# ============================================================
# 3. METRICS
# ============================================================

def compute_metrics(eval_pred):

    logits = eval_pred.predictions
    labels = eval_pred.label_ids

    predictions = np.argmax(logits, axis=-1)

    accuracy = accuracy_score(
        labels,
        predictions
    )

    precision, recall, f1, _ = precision_recall_fscore_support(
        labels,
        predictions,
        average="weighted",
        zero_division=0
    )

    return {
        "accuracy": accuracy,
        "precision": precision,
        "recall": recall,
        "f1": f1
    }


# ============================================================
# 4. TRAINING CONFIGURATION
# ============================================================

training_args = TrainingArguments(

    output_dir="/content/crisismmd_distilbert",

    num_train_epochs=1,

    per_device_train_batch_size=16,

    per_device_eval_batch_size=32,

    learning_rate=5e-5,

    eval_strategy="epoch",

    save_strategy="epoch",

    save_total_limit=1,

    load_best_model_at_end=True,

    metric_for_best_model="f1",

    greater_is_better=True,

    fp16=True,

    logging_steps=100,

    report_to="none"
)


# ============================================================
# 5. CREATE TRAINER
# ============================================================

trainer = Trainer(

    model=model,

    args=training_args,

    train_dataset=train_data,

    eval_dataset=val_data,

    compute_metrics=compute_metrics
)


# ============================================================
# 6. START TRAINING
# ============================================================

print("\n============================================")
print("STARTING CRISISMMD DISTILBERT TRAINING")
print("============================================")

print("GPU:", torch.cuda.get_device_name(0))
print("Train samples:", len(train_data))
print("Validation samples:", len(val_data))
print("Number of classes:", 8)

print("============================================")


train_result = trainer.train()


# ============================================================
# 7. TRAINING COMPLETE
# ============================================================

print("\n============================================")
print("CRISISMMD TRAINING COMPLETED")
print("============================================")

print(
    "Training loss:",
    train_result.training_loss
)
```

```python
# ============================================================
# STEP 9: CRISISMMd RISK + PRIORITY SCORING
# ============================================================

import pandas as pd
import numpy as np

print("============================================")
print("CREATING RISK AND PRIORITY SCORING SYSTEM")
print("============================================")

# Risk weights based on humanitarian urgency
RISK_WEIGHTS = {
    "injured_or_dead_people": 100,
    "missing_or_found_people": 95,
    "affected_individuals": 85,
    "infrastructure_and_utility_damage": 75,
    "rescue_volunteering_or_donation_effort": 65,
    "vehicle_damage": 60,
    "other_relevant_information": 40,
    "not_humanitarian": 10
}

# Priority levels
def get_priority(score):
    if score >= 90:
        return "P1 - CRITICAL"
    elif score >= 75:
        return "P2 - HIGH"
    elif score >= 50:
        return "P3 - MEDIUM"
    else:
        return "P4 - LOW"


# Recommended response
ACTIONS = {
    "injured_or_dead_people":
        "Immediate medical and rescue response",

    "missing_or_found_people":
        "Initiate search and rescue operation",

    "affected_individuals":
        "Provide emergency shelter, food and essential supplies",

    "infrastructure_and_utility_damage":
        "Deploy infrastructure and utility repair teams",

    "rescue_volunteering_or_donation_effort":
        "Coordinate volunteers, donations and relief resources",

    "vehicle_damage":
        "Deploy transportation and road assistance",

    "other_relevant_information":
        "Monitor and assess information",

    "not_humanitarian":
        "No immediate humanitarian response required"
}


def calculate_risk(label, confidence):

    base_score = RISK_WEIGHTS.get(label, 30)

    # Confidence ranges from 0 to 1
    confidence_factor = 0.5 + (0.5 * confidence)

    risk_score = base_score * confidence_factor

    return round(min(risk_score, 100), 2)


print("Risk scoring rules created.")
print("Priority rules created.")
print("Response recommendations created.")

print("\nRisk weights:")
for label, score in RISK_WEIGHTS.items():
    print(f"{label:45s} -> {score}")

print("\n============================================")
print("RISK SCORING SYSTEM READY")
print("============================================")
```

```python
# ============================================================
# STEP 10: GENERATE PREDICTIONS + RISK + PRIORITY
# ============================================================

import numpy as np
import pandas as pd

print("============================================")
print("GENERATING CRISIS RESPONSE PREDICTIONS")
print("============================================")

# ------------------------------------------------------------
# 1. Get predictions from the already-trained model
# ------------------------------------------------------------

prediction_output = trainer.predict(test_dataset)

logits = prediction_output.predictions

# ------------------------------------------------------------
# 2. Convert logits to probabilities
# ------------------------------------------------------------

exp_logits = np.exp(
    logits - np.max(logits, axis=1, keepdims=True)
)

probabilities = exp_logits / exp_logits.sum(
    axis=1,
    keepdims=True
)

# ------------------------------------------------------------
# 3. Predicted class and confidence
# ------------------------------------------------------------

predicted_labels = np.argmax(
    probabilities,
    axis=1
)

confidence_scores = np.max(
    probabilities,
    axis=1
)

# ------------------------------------------------------------
# 4. Label mapping
# ------------------------------------------------------------

label_mapping = {
    0: "affected_individuals",
    1: "infrastructure_and_utility_damage",
    2: "injured_or_dead_people",
    3: "missing_or_found_people",
    4: "not_humanitarian",
    5: "other_relevant_information",
    6: "rescue_volunteering_or_donation_effort",
    7: "vehicle_damage"
}

# ------------------------------------------------------------
# 5. Create result dataframe
# ------------------------------------------------------------

results_df = test_crisis.reset_index(drop=True).copy()

results_df["predicted_label"] = predicted_labels

results_df["confidence"] = confidence_scores

results_df["predicted_category"] = results_df[
    "predicted_label"
].map(label_mapping)

# ------------------------------------------------------------
# 6. Calculate risk score
# ------------------------------------------------------------

results_df["risk_score"] = results_df.apply(
    lambda row: calculate_risk(
        row["predicted_category"],
        row["confidence"]
    ),
    axis=1
)

# ------------------------------------------------------------
# 7. Calculate priority
# ------------------------------------------------------------

results_df["priority"] = results_df[
    "risk_score"
].apply(get_priority)

# ------------------------------------------------------------
# 8. Recommended response
# ------------------------------------------------------------

results_df["recommended_action"] = results_df[
    "predicted_category"
].map(ACTIONS)

# ------------------------------------------------------------
# 9. Display results
# ------------------------------------------------------------

print("\n============================================")
print("PREDICTION PIPELINE COMPLETED")
print("============================================")

print("Total predictions:", len(results_df))

print("\nSample predictions:")

display(
    results_df[
        [
            "tweet_text",
            "predicted_category",
            "confidence",
            "risk_score",
            "priority",
            "recommended_action"
        ]
    ].head(10)
)
```

```python
# ============================================================
# STEP 11: RISK & PRIORITY ANALYSIS
# ============================================================

print("============================================")
print("CRISIS RESPONSE RISK ANALYSIS")
print("============================================")

# ------------------------------------------------------------
# 1. Priority distribution
# ------------------------------------------------------------

print("\nPRIORITY DISTRIBUTION")
print("--------------------------------------------")

priority_distribution = (
    results_df["priority"]
    .value_counts()
    .reindex(
        [
            "P1 - CRITICAL",
            "P2 - HIGH",
            "P3 - MEDIUM",
            "P4 - LOW"
        ],
        fill_value=0
    )
)

display(
    priority_distribution.to_frame(
        name="Number of Reports"
    )
)

# ------------------------------------------------------------
# 2. Category distribution
# ------------------------------------------------------------

print("\nPREDICTED HUMANITARIAN CATEGORIES")
print("--------------------------------------------")

category_distribution = (
    results_df["predicted_category"]
    .value_counts()
)

display(
    category_distribution.to_frame(
        name="Number of Reports"
    )
)

# ------------------------------------------------------------
# 3. Risk statistics
# ------------------------------------------------------------

print("\nRISK SCORE STATISTICS")
print("--------------------------------------------")

print(
    "Average Risk Score:",
    round(results_df["risk_score"].mean(), 2)
)

print(
    "Maximum Risk Score:",
    round(results_df["risk_score"].max(), 2)
)

print(
    "Minimum Risk Score:",
    round(results_df["risk_score"].min(), 2)
)

# ------------------------------------------------------------
# 4. High-risk reports
# ------------------------------------------------------------

print("\nTOP 10 HIGHEST-RISK REPORTS")
print("--------------------------------------------")

top_risk = results_df.sort_values(
    "risk_score",
    ascending=False
).head(10)

display(
    top_risk[
        [
            "tweet_text",
            "predicted_category",
            "confidence",
            "risk_score",
            "priority",
            "recommended_action"
        ]
    ]
)

print("\n============================================")
print("RISK ANALYSIS COMPLETED")
print("============================================")
```

```python
# ============================================================
# DATASET 2 — CRISISMMd
# RECREATE PREDICTIONS + RISK + PRIORITY
# ============================================================

import os
import numpy as np
import pandas as pd
import torch
from torch.nn.functional import softmax

print("============================================")
print("RECREATING CRISISMMd PREDICTIONS")
print("============================================")

# ------------------------------------------------------------
# 1. CHECK REQUIRED OBJECTS
# ------------------------------------------------------------

print("Checking required objects...")

print("Model:", type(model))
print("Test dataset:", len(test_dataset))

# ------------------------------------------------------------
# 2. LABEL MAPPING
# ------------------------------------------------------------

label_names = {
    0: "affected_individuals",
    1: "infrastructure_and_utility_damage",
    2: "injured_or_dead_people",
    3: "missing_or_found_people",
    4: "not_humanitarian",
    5: "other_relevant_information",
    6: "rescue_volunteering_or_donation_effort",
    7: "vehicle_damage"
}

# ------------------------------------------------------------
# 3. RISK WEIGHTS
# ------------------------------------------------------------

risk_weights = {
    "injured_or_dead_people": 100,
    "missing_or_found_people": 95,
    "affected_individuals": 85,
    "infrastructure_and_utility_damage": 75,
    "rescue_volunteering_or_donation_effort": 65,
    "vehicle_damage": 60,
    "other_relevant_information": 40,
    "not_humanitarian": 10
}

# ------------------------------------------------------------
# 4. PRIORITY FUNCTION
# ------------------------------------------------------------

def get_priority(score):

    if score >= 80:
        return "P1 - CRITICAL"

    elif score >= 60:
        return "P2 - HIGH"

    elif score >= 40:
        return "P3 - MEDIUM"

    else:
        return "P4 - LOW"


# ------------------------------------------------------------
# 5. RESPONSE RECOMMENDATION
# ------------------------------------------------------------

def get_action(category):

    actions = {

        "injured_or_dead_people":
            "Immediate medical and rescue response",

        "missing_or_found_people":
            "Immediate search and rescue operation",

        "affected_individuals":
            "Provide emergency shelter, food and essential supplies",

        "infrastructure_and_utility_damage":
            "Deploy infrastructure repair and utility restoration teams",

        "rescue_volunteering_or_donation_effort":
            "Coordinate volunteers, donations and relief resources",

        "vehicle_damage":
            "Coordinate transportation and recovery support",

        "other_relevant_information":
            "Monitor and assess information",

        "not_humanitarian":
            "No immediate humanitarian response required"
    }

    return actions.get(
        category,
        "Monitor and assess information"
    )


# ------------------------------------------------------------
# 6. RUN MODEL ON TEST DATA
# ------------------------------------------------------------

model.eval()
model.to(device)

all_predictions = []
all_confidences = []

print("\nRunning model predictions...")

with torch.no_grad():

    for i in range(len(test_dataset)):

        item = test_dataset[i]

        input_ids = torch.tensor(
            item["input_ids"]
        ).unsqueeze(0).to(device)

        attention_mask = torch.tensor(
            item["attention_mask"]
        ).unsqueeze(0).to(device)

        outputs = model(
            input_ids=input_ids,
            attention_mask=attention_mask
        )

        probabilities = softmax(
            outputs.logits,
            dim=1
        )

        confidence, prediction = torch.max(
            probabilities,
            dim=1
        )

        all_predictions.append(
            prediction.item()
        )

        all_confidences.append(
            confidence.item()
        )

print("Predictions completed:", len(all_predictions))


# ------------------------------------------------------------
# 7. GET ORIGINAL TEST TEXT
# ------------------------------------------------------------

if "tweet_text" in test_dataset.column_names:

    texts = test_dataset["tweet_text"]

else:

    texts = test_crisis["tweet_text"].tolist()


# ------------------------------------------------------------
# 8. CREATE PREDICTION DATAFRAME
# ------------------------------------------------------------

predicted_categories = [
    label_names[p]
    for p in all_predictions
]

risk_scores = []

priorities = []

recommended_actions = []

for category, confidence in zip(
    predicted_categories,
    all_confidences
):

    base_risk = risk_weights[category]

    # Confidence-adjusted risk
    risk = base_risk * confidence

    risk_scores.append(round(risk, 2))

    priorities.append(
        get_priority(risk)
    )

    recommended_actions.append(
        get_action(category)
    )


predictions_df = pd.DataFrame({

    "tweet_text": texts,

    "predicted_category":
        predicted_categories,

    "confidence":
        [round(x, 4) for x in all_confidences],

    "risk_score":
        risk_scores,

    "priority":
        priorities,

    "recommended_action":
        recommended_actions
})


# ------------------------------------------------------------
# 9. DISPLAY RESULTS
# ------------------------------------------------------------

print("\n============================================")
print("PREDICTION PIPELINE COMPLETED")
print("============================================")

print(
    "Total predictions:",
    len(predictions_df)
)

display(
    predictions_df.head(10)
)
```
